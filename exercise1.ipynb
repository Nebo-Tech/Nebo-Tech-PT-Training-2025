```python
import datetime

# Get current date and time
current_datetime = datetime.datetime.now()

# Display result
print("Current date and time :")
print(current_datetime.strftime("%Y-%m-%d %H:%M:%S"))
```

```python
import math

# Prompt user to enter the radius
radius = float(input("Enter the radius of the circle: "))

# Calculate the area using the formula: Area = π * r^2
area = math.pi * radius ** 2

# Display the result
print("r =", radius)
print("Area =", area)

```

```python
# Ask the user to input their first and last names
first_name = input("Enter your first name: ")
last_name = input("Enter your last name: ")

# Print the names in reverse order
print(last_name, first_name)

```

```python
# Accept input from the user
input_values = input("Enter comma-separated numbers: ")

# Generate list and tuple from the input
value_list = input_values.split(",")
value_tuple = tuple(value_list)

# Display the results
print("List :", value_list)
print("Tuple :", value_tuple)


```
