```python
# Define the range
start = 25
end = 50

print("Prime numbers between", start, "and", end, "are:")

# Loop through each number in the range
for num in range(start, end + 1):
    # Prime numbers are greater than 1
    if num > 1:
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                break
        else:
            print(num)
```

```python
# Function to flatten a nested list
def flatten_list(nested_list):
    flattened = []
    for item in nested_list:
        if isinstance(item, list):
            for element in item:
                flattened.append(element)
        else:
            flattened.append(item)
    return flattened

# Test data
nested_list = [1, [2, 3], [4, 5, 6], 7, [8, 9]]
flattened = flatten_list(nested_list)
print("Flattened list:", flattened)
```

```python
def find_max_min_digits(number):
    # Convert to string and filter only digit characters
    digits = [int(d) for d in str(abs(number))]
    
    # Find max and min
    largest = max(digits)
    smallest = min(digits)
    
    # Print results
    print(f"Largest digit in {number}: {largest}")
    print(f"Smallest digit in {number}: {smallest}")
    
# Test cases
num1 = 9876543210
num2 = -5082

find_max_min_digits(num1)
print()
find_max_min_digits(num2)
```
